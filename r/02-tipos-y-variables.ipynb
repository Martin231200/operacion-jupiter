# Tipos de datos
## Enteros y reales
En R, los números se manejan por defecto como "double" (reales de doble precisión); para forzar un valor a tipo entero se agrega el sufijo L. La función class() indica la clase de un objeto y typeof() su tipo interno de almacenamiento.

```r
#Real (double) por defecto
x <- 5
class(x)
#Entero explícito
y <- 5L
class(y)
is.integer(y)
```

## Booleanos
El tipo lógico (logical) toma los valores TRUE y FALSE (o sus abreviaturas T y F). Se generan al evaluar comparaciones y se combinan con los operadores & (y), | (o) y ! (negación); las versiones && y || evalúan un solo elemento y se usan en condicionales.

```r
a <- TRUE
b <- FALSE
a & b
a | b
!a
5 > 3
```

## Strings
El tipo character almacena cadenas de texto entre comillas simples o dobles. nchar() devuelve la longitud de la cadena y paste() o paste0() concatenan cadenas.

```r
s <- "Operación Jupyter"
class(s)
nchar(s)
s2 <- paste("Hola", "mundo")
```

# Declaración de variables
R no requiere declarar el tipo de una variable: el tipo se infiere del valor asignado y puede cambiar si se reasigna. El operador de asignación preferido es <-, aunque = también funciona dentro de una instrucción; -> asigna de izquierda a derecha. Los nombres de variables deben empezar con una letra o un punto (no seguido de número) y distinguen mayúsculas de minúsculas.

```r
x <- 10
x = 10
10 -> x
```
