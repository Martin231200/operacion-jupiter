## Enteros y reales

## Booleanos
El tipo lógico (logical) toma los valores TRUE y FALSE (o sus abreviaturas T y F). Se generan al evaluar comparaciones y se combinan con los operadores & (y), | (o) y ! (negación); las versiones && y || evalúan un solo elemento y se usan en condicionales.

```r
a <- TRUE
b <- FALSE
a & b
a | b
!a
5 > 3
```

## Strings
El tipo character almacena cadenas de texto entre comillas simples o dobles. nchar() devuelve la longitud de la cadena y paste() o paste0() concatenan cadenas.

```r
s <- "Operación Jupyter"
class(s)
nchar(s)
s2 <- paste("Hola", "mundo")
```

# Declaración de variables

R no requiere declarar el tipo de una variable: el tipo se infiere del valor asignado y puede cambiar si se reasigna. El operador de asignación preferido es `<-`; `->` asigna de izquierda a derecha. `=` también asigna, pero **solo a nivel superior** (fuera de una llamada a función) — **dentro** de una llamada, `f(x = 3)`, el `=` no asigna: nombra un argumento de esa función. Los nombres de variables deben empezar con una letra o un punto (no seguido de número) y distinguen mayúsculas de minúsculas.

```r
x <- 10
x = 10
10 -> x
```

```r
mi_variable.1 <- 3.14
```

## Ejemplo: `=` fuera y dentro de una llamada

```r
x = 5        # asignación, equivalente a x <- 5
f <- function(x) x * 2
f(x = 3)     # aquí x = 3 nombra el argumento x de f, no crea/reasigna una variable x
```

## Profundizando en este tema

**Los tipos básicos de R.** Un valor en R es de uno de estos tipos: numérico (`numeric`, con decimales), entero (`integer`, que se escribe con una L: `5L`), texto (`character`), lógico (`logical`: `TRUE` o `FALSE`) y el valor especial `NULL`, que significa «nada». Con `class(x)` y `typeof(x)` puedes preguntarle a R qué tipo tiene un dato; verás que `5` es numérico (de doble precisión) y `5L` es entero.

**Valores faltantes: `NA`.** En estadística es normal que falte un dato. R los representa con `NA`, y casi cualquier cálculo con un `NA` da `NA`: la media de un vector que lo contiene también es `NA`. Muchas funciones aceptan `na.rm = TRUE` para ignorarlos, por ejemplo `mean(x, na.rm = TRUE)`.

**Coerción automática.** Un vector solo puede guardar un tipo. Si mezclas, R convierte al tipo más general: `c(1, "a")` produce `"1" "a"`, todo texto. Puedes convertir a propósito con `as.numeric()`, `as.character()` o `as.integer()`.

**Operadores.** La potencia es `^`, el residuo es `%%` y la división entera es `%/%`. Y como en cualquier lenguaje, `0.1 + 0.2 == 0.3` da `FALSE`; para comparar decimales usa `isTRUE(all.equal(a, b))`.

## Ejercicios

Tres problemas cortos aplicando lo visto, con contexto de ingeniería.

### 1. Conversión de temperatura <span class="badge-dificultad basico">Básico</span> (reales)
Convierte 23.5 °C a Fahrenheit con `F = C * 9/5 + 32`.

```r
c <- 23.5
f <- c * 9/5 + 32

stopifnot(f == 74.3)
cat(c, "°C =", f, "°F\n")
```

### 2. Verificar tolerancia de una pieza <span class="badge-dificultad basico">Básico</span> (booleano)
Una pieza pesa 4.97 kg; el nominal es 5.0 kg con tolerancia ±0.05 kg. ¿Pasa la prueba?

```r
peso <- 4.97
nominal <- 5.0
tolerancia <- 0.05

pasa <- abs(peso - nominal) <= tolerancia
stopifnot(pasa == TRUE)
cat("¿Pasa la prueba?", pasa, "\n")
```

### 3. Etiqueta de producto <span class="badge-dificultad intermedio">Intermedio</span> (strings)
Arma la etiqueta `CÓDIGO-LOTE-NÚMERO` (número con 4 dígitos) para código `"ENG"`, lote `2026`, pieza `47`.

```r
codigo <- "ENG"
lote <- 2026
numero_pieza <- 47

etiqueta <- sprintf("%s-%d-%04d", codigo, lote, numero_pieza)
stopifnot(etiqueta == "ENG-2026-0047")
print(etiqueta)
```
