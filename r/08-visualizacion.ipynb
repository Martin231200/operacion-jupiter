# Visualización

Los gráficos básicos: una línea (tendencia continua) y barras (comparar categorías).

## Gráfica de línea

```r
x <- seq(0, 10, length.out = 100)
y <- sin(x)

plot(x, y, type = "l", col = "#378ADD", lwd = 2,
     main = "Onda senoidal", xlab = "x", ylab = "sin(x)")
grid()
```

## Gráfica de barras

```r
categorias <- c("A", "B", "C", "D")
valores <- c(23, 45, 12, 38)

barplot(valores, names.arg = categorias, col = "#378ADD",
        main = "Comparación por categoría", ylab = "Valor")
```
