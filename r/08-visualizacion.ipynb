## Visualización en R

R tiene dos grandes formas de graficar:

- **Gráficos base** (`plot()`, `hist()`, `boxplot()`): vienen incluidos, sin instalar nada.
- **`ggplot2`**: parte del `tidyverse`, se instala con `install.packages("ggplot2")`, estándar para gráficos elaborados. Se construye por capas con `+`.

Para empezar, los gráficos base bastan; `ggplot2` conviene cuando el resultado tiene que verse bien o se repite mucho.

## Gráfica de línea

```r
x <- seq(0, 10, length.out = 100)
y <- sin(x)

plot(x, y, type = "l", col = "#378ADD", lwd = 2,
     main = "Onda senoidal", xlab = "x", ylab = "sin(x)")
grid()
```

## Gráfica de barras

```r
categorias <- c("A", "B", "C", "D")
valores <- c(23, 45, 12, 38)

barplot(valores, names.arg = categorias, col = "#378ADD",
        main = "Comparación por categoría", ylab = "Valor")
```
