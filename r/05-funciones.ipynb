# Funciones
Las funciones se definen con function(), pueden tener argumentos con valores por defecto, y devuelven el valor de la última expresión evaluada (o el indicado con return()).

A diferencia de Python, R permite **argumentos con nombre** en cualquier orden al llamar la función, y **valores por defecto**:

```r
calcular_area <- function(base, altura = 1) {
  base * altura
}

calcular_area(altura = 5, base = 3)  # el orden no importa si usas nombres
calcular_area(3)                      # usa altura = 1 por defecto
```

R también soporta funciones anónimas (sin nombre), útiles para pasarlas como argumento, por ejemplo dentro de `sapply()`:

```r
sapply(1:5, function(x) x^2)
```

````r
suma <- function(a, b = 0) {
return(a + b)
}
suma(3, 4)
suma(3)
````

## Ejercicios

Tres problemas enfocados en definir funciones propias, con contexto de ingeniería.

### 1. Área de una sección circular <span class="badge-dificultad basico">Básico</span> (valor por defecto)
Define `area_circular(radio, decimales=2)` que regrese el área redondeada. Prueba con radio 5.

````r
area_circular <- function(radio, decimales = 2) {
  round(pi * radio^2, decimales)
}

resultado <- area_circular(5)
stopifnot(resultado == 78.54)
print(resultado)
````

### 2. Conversión con múltiples retornos <span class="badge-dificultad intermedio">Intermedio</span>
Define `convertir_metros(m)` que regrese una lista con `pies` y `pulgadas`. Prueba con 10 metros.

````r
convertir_metros <- function(m) {
  pies <- round(m * 3.28084, 2)
  pulgadas <- round(m * 39.3701, 2)
  list(pies = pies, pulgadas = pulgadas)
}

resultado <- convertir_metros(10)
stopifnot(resultado$pies == 32.81, resultado$pulgadas == 393.7)
print(resultado)
````

### 3. Validación antes de calcular <span class="badge-dificultad intermedio">Intermedio</span> (raíz cuadrada segura)
Define `raiz_segura(x)` que regrese `NA` si `x` es negativo. Prueba con 16 y -4.

````r
raiz_segura <- function(x) {
  if (x < 0) return(NA)
  sqrt(x)
}

r1 <- raiz_segura(16)
r2 <- raiz_segura(-4)
stopifnot(r1 == 4, is.na(r2))
cat(r1, r2, "\n")
````
