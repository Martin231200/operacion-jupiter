# Particularidades de cada lenguaje
## R
-   Indexado desde 1 (no desde 0).
-   Vectorizado por diseño: las operaciones aritméticas se aplican elemento a elemento sin necesidad de bucles explícitos.
-   Orientado a estadística y visualización de datos; ecosistema CRAN muy amplio para análisis estadístico.
-   Asignación con <- como convención, aunque = es válido.
## Python
-   Indexado desde 0.
-   Lenguaje de propósito general; su uso en ciencia de datos depende de bibliotecas externas (NumPy, pandas, Matplotlib).
-   Sintaxis basada en indentación en vez de llaves para delimitar bloques.
-   Tipado dinámico, igual que R, pero con una sintaxis orientada a objetos más explícita (todo es un objeto con métodos).
## Julia
-   Indexado desde 1, como R.
-   Diseñado para desempeño numérico: se compila "just-in-time" (JIT), por lo que puede acercarse a la velocidad de C manteniendo una sintaxis de alto nivel.
-   Nació explícitamente para resolver el "problema de los dos lenguajes" (prototipar en un lenguaje interpretado y reescribir en uno compilado para producción).
-   Soporte nativo para multiple dispatch: el comportamiento de una función puede depender del tipo de todos sus argumentos, no solo del primero.
Nota: las particularidades de Julia se documentan aquí con base en las características generales del lenguaje; al no contar todavía con el notebook de Julia del proyecto, conviene revisarlas junto con el material específico cuando esté disponible.
# Conclusiones
Los tres lenguajes permiten resolver problemas similares de programación y análisis de datos, pero parten de objetivos distintos: R nació para estadística y gráficos, Python como lenguaje general que se volvió estándar en ciencia de datos gracias a sus bibliotecas, y Julia como respuesta al costo de desempeño de los otros dos en cómputo numérico intensivo. Para quien ya conoce uno de los tres, la transición a los otros es principalmente de sintaxis: la lógica de secuencias, condicionales, ciclos y funciones es la misma, y las diferencias reales aparecen en el manejo de estructuras de datos (vectorizado en R y Julia, basado en bibliotecas en Python) y en el ecosistema de cada uno.
