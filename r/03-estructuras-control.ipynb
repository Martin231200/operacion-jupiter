# Estructuras: secuencias, decisiones y repeticiones
## Secuencias
El operador : genera una secuencia de enteros consecutivos; la función seq() permite controlar el incremento y seq_len()/seq_along() generan secuencias a partir de la longitud de un objeto.
1:10

```r
seq(0, 1, by = 0.25)
seq_len(5)
```

## Estructura "if"

```r
if (x > 0) {
print("positivo")
}
```

## Estructura "if-else"

```r
if (x > 0) {
print("positivo")
} else {
print("no positivo")
}
```

## Estructura "for"

```r
for (i in 1:5) {
print(i)
}
```

## Estructura "while"

```r
i <- 1
while (i <= 5) {
print(i)
i <- i + 1
}
```

## Estructura "repeat"
repeat ejecuta el bloque indefinidamente hasta encontrar un break; es el equivalente en R de un bucle "do-while".

```r
i <- 1
repeat {
print(i)
i <- i + 1
if (i > 5) break
}
```
