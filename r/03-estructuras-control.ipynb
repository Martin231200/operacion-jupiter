## Secuencias
El operador : genera una secuencia de enteros consecutivos; la función seq() permite controlar el incremento y seq_len()/seq_along() generan secuencias a partir de la longitud de un objeto.
1:10

```r
seq(0, 1, by = 0.25)
seq_len(5)
```

#### Estructura "if"

```r
if (x > 0) {
print("positivo")
}
```

#### Estructura "if-else"

```r
if (x > 0) {
print("positivo")
} else {
print("no positivo")
}
```

#### Estructura "for"

```r
for (i in 1:5) {
print(i)
}
```

#### Estructura "while"

```r
i <- 1
while (i <= 5) {
print(i)
i <- i + 1
}
```

#### Estructura "repeat"

repeat ejecuta el bloque indefinidamente hasta encontrar un break; es el equivalente en R de un bucle "do-while".

```r
i <- 1
repeat {
print(i)
i <- i + 1
if (i > 5) break
}
```

## Ejercicios

Tres problemas cortos con estructuras de control, contexto de ingeniería.

### 1. Consumo total de energía (ciclo `for`)
Suma el consumo (kWh) de 5 máquinas: `c(120, 95, 150, 80, 110)`.

```r
consumos <- c(120, 95, 150, 80, 110)
total <- 0
for (c in consumos) {
  total <- total + c
}

stopifnot(total == 555)
cat("Consumo total:", total, "kWh\n")
```

### 2. Clasificar estado por temperatura (`if`/`else if`/`else`)
Clasifica una máquina: `"normal"` si temp < 60, `"alerta"` si < 85, `"critico"` en otro caso. Prueba con 45, 70 y 90 °C.

```r
clasificar <- function(temp) {
  if (temp < 60) {
    return("normal")
  } else if (temp < 85) {
    return("alerta")
  } else {
    return("critico")
  }
}

resultados <- sapply(c(45, 70, 90), clasificar)
stopifnot(all(resultados == c("normal", "alerta", "critico")))
print(resultados)
```

### 3. Simular llenado de un tanque (`while`)
Un tanque se llena a 45 L/min hasta llegar a 500 L. ¿Cuántos minutos completos tarda en alcanzar o superar el objetivo?

```r
nivel <- 0
objetivo <- 500
tasa <- 45
minutos <- 0

while (nivel < objetivo) {
  nivel <- nivel + tasa
  minutos <- minutos + 1
}

stopifnot(minutos == 12)
cat("Minutos:", minutos, ", nivel final:", nivel, "L\n")
```
