# Lenguajes de programación de código libre
## R y RStudio
### Características
-   R es un lenguaje interpretado, lo que significa que los usuarios acceden a sus funciones a través de un intérprete de línea de comandos
-   A diferencia de Python y Java, R es un lenguaje de programación específico de dominio (diseñado para un campo en especial, en este caso orientado al cálculo estadístico y generación de gráficos)
-   Es un software libre, lo que significa que no tiene costo en su descarga, uso y distribución e incluso modificación
-   R es competencia de Excel, Phyton, SPSS y STATA
-   Se parece a Matlab y su sintaxis recuerda a C / C++
-   R fue desarrollado inicialmente por Ross Ihaka y Robert Gentleman en 1995 dentro de la Universidad de Auckland, Nueva Zelanda, y se encuentra en desarrollo constante por el R Development Core Team
-   La versión beta más estable fue lanzada hasta el año 2000
-   RStudio es un entorno de desarrollo integrado (IDE) para R
### Ventajas y desventajas de R
-   Es un software libre
-   No hay responsabilidad en caso de errores en el código
-   Es uno de los primeros lenguajes en recibir los nuevos temas
-   Tiene una fácil conectividad y compatibilidad con otros lenguajes de programación
-   No tiene límite (teórico) en la lectura de bases de datos
-   Suele volverse más lento el análisis con grandes bases de datos
# Interfases y ambientes colaborativos
## Correr en R o RStudio
Las ventajas de RStudio en comparación con R son principalmente que se tiene un ambiente integrado, en donde se marcan los errores y que las diferentes ventanas permiten tener un horizonte completo del estado de la simulación.
Para instalar R y RStudio se siguen los siguientes pasos:
-   Primero se debe instalar R
-   Dependiendo de las características del sistema operativo, se puede encontrar en https://www.r-project.org/ el archivo de instalación adecuado
-   Después de elegir el idioma, acepte los términos
-   Al final de la instalación busque la carpeta en donde se guardó (generalmente *Aplicaciones*) para inicarlo
-   Para facilitar el proceso de información y bibliotecas es conveniente utilizar RStudio cuya interacción es más amigable. Es un entorno de desarrollo integrado (IDE, por sus siglas en inglés) que con la ayuda de una consola, un editor de código intuitivo, herramientas especiales de edición, visor de gráficos, entre otros, permite programar en R de una forma fácil y cómoda.
-   Se descarga RStudio en https://www.rstudio.com/, actualmente disponible en https://posit.co/, en donde se va a la pestaña "Open Source", de donde se puede descargar.

## Ventajas de las diferentes interfaces
Las ventajas de RStudio en comparación con Google Colab son:
-   Se tiene en cada momento un registro de las variables definidos y sus valores
-   Se puede correr el código de línea en línea, o incluso partes de una línea (esto mismo se puede hacer en Colab, seleccionando la línea y picando CTRL-SHIFT-ENTER en Windows o COMMAND-SHIFT-ENTER en Mac)
Ventajas de correr en Google Colab:
-   El código desarrollado es fácilmente intercambiable con otros usuarios para trabajo colaborativo
-   No hace falta instalar R en la computadora por lo que es más flexible (por ejemplo, para uso en una tablet, para uso en clase, \...)
Desventajas de correr en Google Colab:
-   Aunque el sistema está diseñado para Python, el sistema permite trabajar en R. Sin embargo, no se puede guardar el archivo resultante en formato .R, por lo que los códigos son poco intercambiables.
-   El IDE tiene menos opciones que en RStudio
-   El sistema Markdown de colab no es el mismo que en RStudio

# Primeros pasos
## Comandos basicos
Se puede encontrar información general sobre la licencia de R, indicando que es un software libre, y sobre las personas que han contribuido al lenguaje, con las siguientes instrucciones:

```r
#Licencia de R
license()
#Contribuyentes
contributors()
#Cita
citation()
```

La ultima línea genera una cita en R, así como la información para generar una entrada BibTex. Si se agrega entre las paréntesis el nombre de un paquete, se genera la cita de ese paquete. Si se deja vacío el espacio, se genera la cita para el lenguaje de programación, en cuyo caso los resultados son:
