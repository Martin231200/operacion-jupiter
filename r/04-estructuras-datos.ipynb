# Estructuras de datos y objetos
## Vectores
El vector es la estructura base de R: una secuencia unidimensional de elementos del mismo tipo, creada con c().

```r
v <- c(1, 2, 3, 4)
```

v\[2\]

```r
length(v)
```

## Factores
Los factores representan variables categóricas; almacenan internamente enteros asociados a etiquetas (niveles).

```r
f <- factor(c("bajo", "alto", "medio"))
levels(f)
```

## Matrices
Una matriz es un vector con dos dimensiones (filas y columnas), y sus elementos deben ser del mismo tipo.

```r
m <- matrix(1:6, nrow = 2, ncol = 3)
dim(m)
```

## Arreglos
array() generaliza la matriz a más de dos dimensiones.

```r
a <- array(1:24, dim = c(2, 3, 4))
dim(a)
```

## Dataframes
El data.frame es la estructura tabular estándar de R: columnas que pueden ser de distinto tipo, análoga a una hoja de cálculo.

```r
df <- data.frame(nombre = c("A", "B"), valor = c(1, 2))
str(df)
```

## Listas
La lista es la estructura más flexible: sus elementos pueden ser de tipos y longitudes distintas, incluyendo otras listas.

```r
lst <- list(numero = 1, texto = "hola", vector = c(1,2,3))
lst\$texto
```
