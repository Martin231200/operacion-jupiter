## Estructuras de datos en R

R tiene cuatro estructuras de datos principales, y elegir la correcta depende de si tus datos son de un solo tipo o mixtos, y de si son de una o más dimensiones:

- **Vector**: secuencia de elementos del mismo tipo. Es la estructura más básica — incluso un solo número es técnicamente un vector de longitud 1.
- **Lista**: como un vector, pero sus elementos pueden ser de tipos distintos e incluso otras listas o data frames.
- **Matriz**: un vector con dos dimensiones (filas y columnas), todos del mismo tipo.
- **Data frame**: una tabla donde cada columna puede ser de un tipo distinto (como una hoja de cálculo).

El acceso a elementos cambia según la estructura: con `[ ]` obtienes un subconjunto conservando la estructura original, y con `[[ ]]` obtienes el elemento individual (importante sobre todo en listas).

# Estructuras de datos y objetos
## Vectores
El vector es la estructura base de R: una secuencia unidimensional de elementos del mismo tipo, creada con c().

```r
v <- c(1, 2, 3, 4)
```

```r
v[2]
```

```r
length(v)
```

:::: {.panel-tabset}

#### Factores

Los factores representan variables categóricas; almacenan internamente enteros asociados a etiquetas (niveles).

```r
f <- factor(c("bajo", "alto", "medio"))
levels(f)
```

#### Matrices

Una matriz es un vector con dos dimensiones (filas y columnas), y sus elementos deben ser del mismo tipo.

```r
m <- matrix(1:6, nrow = 2, ncol = 3)
dim(m)
```

#### Arreglos

array() generaliza la matriz a más de dos dimensiones.

```r
a <- array(1:24, dim = c(2, 3, 4))
dim(a)
```

#### Dataframes

El data.frame es la estructura tabular estándar de R: columnas que pueden ser de distinto tipo, análoga a una hoja de cálculo.

```r
df <- data.frame(nombre = c("A", "B"), valor = c(1, 2))
str(df)
```

#### Listas

La lista es la estructura más flexible: sus elementos pueden ser de tipos y longitudes distintas, incluyendo otras listas.

```r
lst <- list(numero = 1, texto = "hola", vector = c(1,2,3))
lst$texto
```

## Profundizando en este tema

**Cuatro estructuras, cuatro usos.** El **vector** guarda elementos del mismo tipo; la **lista** acepta tipos mezclados e incluso otras listas; la **matriz** es un vector con dos dimensiones; y el **data frame** es la tabla de siempre: columnas de tipos distintos, pero todas con la misma longitud. Casi todo el análisis de datos en R pasa por data frames.

**Índices desde 1.** El primer elemento de un vector es `v[1]`, no `v[0]`. Un índice negativo **excluye**: `v[-1]` es todo menos el primero. Esto es lo opuesto a Python, donde un índice negativo cuenta desde el final.

**Un corchete o dos.** En una lista, `l[1]` devuelve una lista con un elemento, mientras que `l[[1]]` devuelve el elemento en sí. Y `l$nombre` extrae un elemento por su nombre. En un data frame, `df$columna` extrae una columna completa.

**Reciclaje.** Si operas dos vectores de distinta longitud, R repite el más corto: `c(1, 2, 3, 4) + c(10, 20)` da `11 22 13 24`. Es útil, pero cuando las longitudes no son múltiplos, R avisa con una advertencia: no la ignores.

## Ejercicios

Tres problemas cortos, contexto de ingeniería.

### 1. Vector: registrar una nueva medición <span class="badge-dificultad basico">Básico</span>
A `c(12.4, 13.1, 11.9)` agrégale `12.8` con `c()` y confirma que ahora tiene 4 elementos.

```r
mediciones <- c(12.4, 13.1, 11.9)
mediciones <- c(mediciones, 12.8)

stopifnot(length(mediciones) == 4)
print(mediciones)
```

### 2. Lista: consultar inventario por código <span class="badge-dificultad basico">Básico</span>
De `list(P100=25, P200=40, P300=15)`, obtén cuántas piezas hay de `P200`.

```r
inventario <- list(P100 = 25, P200 = 40, P300 = 15)
disponibles <- inventario$P200

stopifnot(disponibles == 40)
print(disponibles)
```

### 3. Data frame: filtrar por columna <span class="badge-dificultad intermedio">Intermedio</span>
De un data frame de piezas, obtén el nombre de la fila cuyo valor es 2.

```r
df <- data.frame(nombre = c("A", "B"), valor = c(1, 2))
resultado <- df$nombre[df$valor == 2]

stopifnot(resultado == "B")
print(resultado)
```
