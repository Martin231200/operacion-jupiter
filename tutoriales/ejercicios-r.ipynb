## 1. Matemáticas: verificar si un número es primo

```r
es_primo <- function(n) {
  if (n < 2) return(FALSE)
  for (i in 2:floor(sqrt(n))) {
    if (n %% i == 0) return(FALSE)
  }
  return(TRUE)
}

casos <- list(`7` = TRUE, `15` = FALSE, `2` = TRUE, `97` = TRUE)
for (n in names(casos)) {
  resultado <- es_primo(as.numeric(n))
  stopifnot(resultado == casos[[n]])
  cat(sprintf("es_primo(%s) = %s  [correcto]\n", n, resultado))
}
```

## 3. Resolver una ecuación cuadrática

```r
resolver_cuadratica <- function(a, b, c) {
  disc <- b^2 - 4*a*c
  x1 <- (-b + sqrt(disc)) / (2*a)
  x2 <- (-b - sqrt(disc)) / (2*a)
  return(c(x1, x2))
}

a <- 1; b <- -3; c <- 2  # x^2 - 3x + 2 = 0 -> raíces 1 y 2
raices <- resolver_cuadratica(a, b, c)

for (x in raices) {
  valor <- a*x^2 + b*x + c
  stopifnot(abs(valor) < 1e-9)
}

cat("Raíces:", raices, "\n")
cat("Ambas raíces satisfacen la ecuación (verificado)\n")
```
