Los mismos ejercicios aparecen en R y Julia para comparar resultados y sintaxis.

## 1. Esfuerzo normal

`σ = F / A` es la definición de **esfuerzo**. La ley de Hooke es otra relación: `σ = E · ε`, entre esfuerzo y deformación.

```python
def esfuerzo(fuerza, area):
    return fuerza / area

print(esfuerzo(5000, 0.002))    # N / m²
```

## 2. Área de un círculo con incertidumbre

Sumar una tolerancia directamente al área no tiene sentido dimensional. Lo correcto es propagar la incertidumbre del radio: `ΔA ≈ 2πr·Δr`.

```python
import math

def area_circular(r):
    return math.pi * r ** 2

def area_con_incertidumbre(r, dr):
    return area_circular(r), 2 * math.pi * r * dr

print(area_con_incertidumbre(0.05, 0.001))
```

## 3. Raíces de una cuadrática

```python
def raices(a, b, c):
    disc = b * b - 4 * a * c
    return (-b + math.sqrt(disc)) / (2 * a), (-b - math.sqrt(disc)) / (2 * a)

print(raices(1, -3, 2))
```

Con discriminante negativo, `math.sqrt` lanza `ValueError: math domain error`. R devuelve `NaN` y Julia lanza `DomainError`, buen punto de comparación.

## 4. Verificar con tolerancia

Con `datos = [19.98, 20.01, 19.99, 20.03, 20.00, 20.02, 19.97, 20.00]`, la media es 20.0. Para comparar números decimales usa tolerancia, no igualdad exacta:

```python
datos = [19.98, 20.01, 19.99, 20.03, 20.00, 20.02, 19.97, 20.00]
media = sum(datos) / len(datos)
assert abs(media - 20.0) < 1e-9
print(media)
```
