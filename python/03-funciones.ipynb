## Funciones definidas por el usuario

Forma general de una función:

$ def <nombredelafuncion> (argumentos):$ \\
$     \quad calculos $ \\
$     \quad return (resultados)$

```python
def saludos():
  print("Saludos gente")
  print("La porra los saluda")
```

```python
saludos()
```

```python
def pide_pi():
  numero_pi = math.pi
  return numero_pi
```

```python
pide_pi()
```

```python
def repetir(palabra, n):
    print(palabra * n)
```

```python
spam = 'Spam, '
repetir(spam, 4)
```

```python
def primeras_dos_lineas():
    repetir(spam, 4)
    repetir(spam, 4)
```

```python
primeras_dos_lineas()
```

```python
def ultimas_tres_lineas():
    repetir(spam, 2)
    print('(Lovely Spam, Wonderful Spam!)')
    repetir(spam, 2)
```

```python
ultimas_tres_lineas()
```

```python
def coro():
    primeras_dos_lineas()
    ultimas_tres_lineas()
```

```python
coro()
```

### Ciclos

La estructura de repetición FOR tiene la siguiente sintaxis:

for $<var>$ in $range(valor):$

```python
for i in range(2):
    print(i)
```

```python
for i in range(2):
    print("Coro", i)
    coro()
    print()
```

```python
def n_coros(n):
    for i in range(n):
        coro()
        print()
```

```python
n_coros(5)
```

### Variables y parámetros locales

Las variables y parámentros locales son aquellas cuya influencia se remita a una función.

```python
def dos_partes(parte1,parte2):
  total = parte1 + parte2
  print(total)
```

```python
parte1='yo '
parte2='y tu '
dos_partes(parte1,parte2)
```

```python
if x > 0:
    print('x is positive')  # Decisión sencilla
```

```python
if x % 2 == 0:
    print('x es par')
else:
    print('x es non')  # Decisión doble
```

```python
if x < y:
    print('x es menor que y')
elif x > y:
    print('x es mayor que y')
else:
    print('x y y son iguales') # Decisiones anidadas
```

también puede escribirse como:

```python
if x < y:
    print('x es menor que y')
else:
    if x > y:
           print('x es mayor que y')
    else:
          print('x y y son iguales') # Decisiones anidadas
```
