## Argumentos con nombre y valores por defecto

```python
def calcular_area(base, altura=1):
    return base * altura

print(calcular_area(base=3, altura=5))   # el orden no importa si usas nombres
print(calcular_area(3))                  # usa altura=1
```

## Ciclos con `range`

```python
for i in range(3):
    print(i)
```

`range(3)` produce 0, 1 y 2: **excluye** el último valor.

## Librerías

```python
import math
print(math.sqrt(16))
```
