# Visualización

Los gráficos básicos: una línea (tendencia continua) y barras (comparar categorías).

## Gráfica de línea

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 100)
y = np.sin(x)

plt.plot(x, y)
plt.title("Onda senoidal")
plt.xlabel("x")
plt.ylabel("sin(x)")
plt.grid(True, alpha=0.3)
plt.show()
```

## Gráfica de barras

```python
categorias = ["A", "B", "C", "D"]
valores = [23, 45, 12, 38]

plt.bar(categorias, valores, color="#378ADD")
plt.title("Comparación por categoría")
plt.ylabel("Valor")
plt.show()
```
