## Visualización en Python

La librería más usada es **Matplotlib**; la mayoría de las demás (Seaborn, pandas `.plot()`) están construidas encima de ella. `plt.show()` es necesario en un script normal; en Jupyter/Colab normalmente no hace falta porque la gráfica se muestra sola debajo de la celda.

## Gráfica de línea

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 100)
y = np.sin(x)

plt.plot(x, y)
plt.title("Onda senoidal")
plt.xlabel("x")
plt.ylabel("sin(x)")
plt.grid(True, alpha=0.3)
plt.show()
```

## Gráfica de barras

```python
categorias = ["A", "B", "C", "D"]
valores = [23, 45, 12, 38]

plt.bar(categorias, valores, color="#378ADD")
plt.title("Comparación por categoría")
plt.ylabel("Valor")
plt.show()
```
