# Python: visualización

**Matplotlib** es la librería base para graficar; Seaborn y `pandas.plot()` se construyen encima de ella.

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4]
y = [1, 4, 9, 16]

plt.plot(x, y, marker="o")
plt.xlabel("x")
plt.ylabel("x al cuadrado")
plt.title("Ejemplo de gráfica de línea")
plt.show()
```

Para gráficas estadísticas (histogramas, cajas, mapas de calor) con menos código, **Seaborn** suele ser más práctico:

```python
import seaborn as sns
import numpy as np

rng = np.random.default_rng(0)
datos = rng.normal(size=200)
sns.histplot(datos)
```

En un script, `plt.show()` abre la ventana de la gráfica. En Jupyter o Colab normalmente no hace falta: la gráfica aparece debajo de la celda.
