# Python: estructuras de datos

Elegir la estructura correcta depende de si necesitas orden, si los elementos pueden repetirse y si vas a modificarlos.

| Estructura | Sintaxis | Mutable | Orden | Duplicados |
|---|---|---|---|---|
| Lista | `[1, 2, 3]` | Sí | Sí | Sí |
| Tupla | `(1, 2, 3)` | No | Sí | Sí |
| Diccionario | `{"a": 1}` | Sí | Orden de inserción | Claves únicas |
| Conjunto | `{1, 2, 3}` | Sí | No | No |

## Listas son mutables

```python
Sopes = ['Pozole', 'Atole', 'Tamales']
Sopes[1] = 'Tasajo'
print(Sopes)
print('Tasajo' in Sopes)
```

## Diccionarios conservan el orden de inserción

Desde Python 3.7, un diccionario recuerda el orden en que agregaste las claves.

```python
d = {"b": 1, "a": 2}
d["c"] = 3
print(list(d))
```

## `min` y `max` también sirven con texto

```python
print(min(["banana", "manzana", "kiwi"]), max([3, 1, 4, 1, 5]))
```

## Ejercicio

Con `inventario = {"tornillos": 120, "tuercas": 340, "arandelas": 85}`, agrega `"remaches": 200`, imprime las claves y encuentra el material con menor cantidad.

```python
inventario = {"tornillos": 120, "tuercas": 340, "arandelas": 85}
inventario["remaches"] = 200
print(list(inventario.keys()))
print(min(inventario, key=inventario.get))
```
