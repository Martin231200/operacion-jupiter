### Métodos

append' agrega al final un elemento a una **Lista**

```python
Sopes.append('Rajas')
Sopes
```

'extend' agrega elementos tomados de una lista adicional

```python
Sopes.extend(['Elote', 'Queso'])
Sopes
```

'pop' elimina el elemento de determinado índice

```python
Sopes.pop(3)
Sopes
```

'remove' elimina el elemento

```python
Sopes.remove('Pollo')
Sopes
```

```python
Sopes.remove('Carne')
```

```python
Nombre = 'Juan'
Nombre
```

'list' divide una cadena en sus elementos

```python
n=list(Nombre)
n
```

'split' en palabras

```python
Ultra='La porra te saluda'
Ultra
```

```python
Ultra.split()
```

En este método se puede establecer el delimitador

```python
f='ex-comadre'
g=f.split('-')
g
```

'join' permite concatenar listas de cadenas en una sola. Se sugiere invocarlo en el delimitador y que la lista sea el argumento

```python
h=['waiting','for','a','girl','like','you']
h
```

```python
delimiter = ' '
j = delimiter.join(h)
j
```

Se puede recorrer una lista con un ciclo contador (for)

```python
for sopes in Sopes:
  print(sopes)
```

```python
for palabra in Ultra.split():
  print(palabra)
```

```python
Sopes
```

## Diccionarios

Un diccionario es una colección mutable no ordenada de pares de datos 'llave:valor' donde la llaves es única pero el valor no necesariamente

```python
mi_dicc = {'nombre':'Juan', 'edad':25 , 'profesión':'ingeniero'} # 'nombre' es la llave, 'Juan' es el valor... Los diccionarios se crean con {}
mi_dicc
```

Para obtener un valor a partir de la llave

```python
print(mi_dicc['nombre'])
```

Para agregar un valor

```python
mi_dicc['ciudad'] = 'Ciudad de México'
mi_dicc
```

Para modificar un valor

```python
mi_dicc['edad'] = 30
mi_dicc
```

Para eliminar un valor

```python
del mi_dicc['profesión']
mi_dicc
```

En un diccionario, el error producido es 'key error'

```python
mi_dicc['llave']
```

```python
len(mi_dicc)
```

```python
mi_dicc
```

```python
'nombre' in mi_dicc #opera sobre la llave
```

## Tuplas

Una tupla es una secuencia de valores de cualquier tipo indexados por números enteros; a diferencia de las listas, son inmutables

```python
z = 'c','o','m','p','u','t','a','c','i','o','n'
z
```

```python
type(z)
```

No es necesario el uso de parentésis

```python
z1 = 'W', # Una tupla de un sólo elemento debe terminar con una ','
type(z1)
```

```python
z2 = 'W' # De lo contrario, es una cadena
type(z2)
```

```python
z3 = tuple('Cadena')
z3
```

```python
z[2]
```

```python
z[1:3] # La mayoría de los operadores que se aplican a listas también se pueden aplicar a tuplas
```

```python
tuple('arr')+tuple('ooooz')
```

```python
tuple('gol')*4
```

```python
sorted(z)
```

```python
reversed(z)
```

```python
tuple(reversed(z))
```

```python
z.remove('o')
```
