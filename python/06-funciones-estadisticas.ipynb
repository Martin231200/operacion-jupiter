# Python: funciones estadísticas

La librería estándar trae el módulo `statistics`; para trabajo serio se usa NumPy.

```python
import numpy as np

m = np.array([20.1, 19.8, 20.3, 19.9, 20.0, 20.2])
print(np.mean(m), np.median(m))
print(round(np.std(m, ddof=1), 4))
```

::: {.callout-important}
## Cuidado con `ddof`
`np.std()` divide entre *n* (`ddof=0`, poblacional). `statistics.stdev()` en Python, `sd()` en R y `std()` en Julia dividen entre *n − 1* (muestral). Con los mismos datos dan resultados distintos; para compararlos, iguala el parámetro con `np.std(x, ddof=1)`.
:::

```python
import statistics
datos = [19.98, 20.01, 19.99, 20.03, 20.00, 20.02, 19.97, 20.00]
print(np.std(datos))                 # ddof=0
print(np.std(datos, ddof=1))         # ddof=1
print(statistics.stdev(datos))       # ddof=1
```
