# Funciones estadísticas

```python
W=np.arange(1,22,2)
print("Arreglo: ", W)
```

```python
print(W.mean())
```

```python
print(W.sum())
```

```python
print(W.min())
```

```python
print(W.max())
```

# Funciones universales

```python
np.square(W)
```

```python
np.sqrt(W)
```

```python
np.exp(W)
```

Numpy aplica operaciones aritméticas sobre arreglos que no tienen la misma forma shape), esa propiedad se llama **broadcasting**

```python
W1=np.arange(5)
W2=np.array([3])
print("Shape W1: ",W1.shape)
print("W1: ",W1)
print()
print("Shape W2: ",W2.shape)
print("W2: ",W2)
```

```python
W1 + W2
```
