# Python: entrada de datos y operadores

## Entrada de datos

`input()` siempre devuelve texto (`str`), aunque la persona escriba un número. Si vas a calcular con él, conviértelo:

```python
edad_texto = "20"     # así llegaría de input("¿Cuántos años tienes? ")
print(type(edad_texto))
edad = int(edad_texto)   # conversión explícita
print(type(edad), edad)
```

## Operadores

| Familia | Operadores | Notas |
|---|---|---|
| Aritméticos | `+  -  *  /  //  %  **` | `//` división entera, `%` residuo, `**` potencia |
| Comparación | `==  !=  >  <  >=  <=` | Devuelven `True` o `False` |
| Lógicos | `and`  `or`  `not` | Palabras completas, no `&&` ni `||` |

```python
print(7 / 2, 7 // 2, 7 % 2, 2 ** 3)
print(3 > 2 and 5 == 5)
```

::: {.callout-warning}
`=` asigna un valor y `==` compara. Confundirlos es uno de los errores más comunes al empezar.
:::
