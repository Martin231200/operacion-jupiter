## Funciones definidas por el usuario

Forma general de una función:

$ def <nombredelafuncion> (argumentos):$ \\
$     \quad calculos $ \\
$     \quad return (resultados)$

```python
def saludos():
  print("Saludos gente")
  print("La porra los saluda")
```

```python
saludos()
```

```python
def pide_pi():
  numero_pi = math.pi
  return numero_pi
```

```python
pide_pi()
```

```python
def repetir(palabra, n):
    print(palabra * n)
```

```python
spam = 'Spam, '
repetir(spam, 4)
```

```python
def primeras_dos_lineas():
    repetir(spam, 4)
    repetir(spam, 4)
```

```python
primeras_dos_lineas()
```

```python
def ultimas_tres_lineas():
    repetir(spam, 2)
    print('(Lovely Spam, Wonderful Spam!)')
    repetir(spam, 2)
```

```python
ultimas_tres_lineas()
```

```python
def coro():
    primeras_dos_lineas()
    ultimas_tres_lineas()
```

```python
coro()
```

### Ciclos

La estructura de repetición FOR tiene la siguiente sintaxis:

for $<var>$ in $range(valor):$

```python
for i in range(2):
    print(i)
```

```python
for i in range(2):
    print("Coro", i)
    coro()
    print()
```

```python
def n_coros(n):
    for i in range(n):
        coro()
        print()
```

## Creación de arreglos
Funciones adicionales para crear un arreglo

```python
# Arreglo cuyos valores son todos 0
a = np.zeros((2,3))
print(a)
```

```python
# Arreglo cuyos valores son todos 1
b = np.ones((2,3,4))
print(b)
```

```python
# Arreglo cuyos valores son todos "8", o cualquier otro valore indicado en el segundo parámentro
c = np.full((3,4),8)
print(c)
```

## Modificación de un arreglo

```python
print("shape",arreglo1.shape)
```

```python
print("arreglo1: ", arreglo1)
```

```python
arreglo1.shape=(4,3)
print("Shape: ", arreglo1.shape)
```
