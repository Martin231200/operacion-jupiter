# Proyecto Jupyter: Aprendamos Python

::: {.callout-note}
## ¿Cómo se organiza un notebook?
Un notebook alterna celdas de código (se ejecutan) y celdas de texto (solo explican). Puedes correr una celda con Shift + Enter.
:::

Python fue creado por Guido van Rossum a finales de los años 80 en el Centrum Wiskunde & Informatica (CWI) en los Países Bajos.

El nombre "Python" proviene de la afición de Guido van Rossum por el grupo de comedia británico Monty Python.

## Instalación

La instalación consiste en descargar el kernel de Python directamente de su página web en la dirección [python.org](https://www.python.org/downloads/) El usuario puede escoger el editor que mejor le convenga.
