## Ejercicios

Tres calentamientos para confirmar que tu entorno funciona.

### 1. Mensaje de bienvenida formateado
Arma el mensaje `"Bienvenido a {proyecto} (v{version})"` con `proyecto="Operación Jupyter"` y `version=1`.

```python
nombre_proyecto = "Operación Jupyter"
version = 1

mensaje = f"Bienvenido a {nombre_proyecto} (v{version})"
assert mensaje == "Bienvenido a Operación Jupyter (v1)"
print(mensaje)
```

### 2. Verificar el entorno con una operación simple
Calcula 2 elevado a la 10 — si tu entorno corre bien, debe dar 1024.

```python
resultado = 2**10
assert resultado == 1024
print(resultado)
```

### 3. Contar caracteres de un mensaje de bitácora
¿Cuántos caracteres tiene `"Sistema iniciado correctamente"`?

```python
mensaje_log = "Sistema iniciado correctamente"
longitud = len(mensaje_log)
assert longitud == 30
print(longitud)
```
