# Python: introducción

Python es un lenguaje de propósito general, de sintaxis legible y con un ecosistema enorme de librerías. Es una buena primera opción para aprender a programar.

## Formas de usar Python

**Instalación local.** Se descarga desde [python.org](https://www.python.org/downloads/) y se usa desde la terminal o desde un editor como VS Code.

**Anaconda.** [Anaconda](https://www.anaconda.com/download) instala Python junto con NumPy, pandas y Matplotlib, además del gestor de entornos `conda`. Es cómoda para ciencia de datos.

**Jupyter.** Con Python ya instalado:

```bash
pip install notebook
jupyter notebook
```

**Google Colab.** Corre notebooks en el navegador sin instalar nada. A diferencia de R y Julia, Python tiene soporte nativo completo.

## Comandos básicos

| Comando | Qué hace |
|---|---|
| `python archivo.py` | Ejecuta un script desde la terminal |
| `python` | Abre el intérprete interactivo (REPL); se sale con `exit()` |
| `pip install paquete` | Instala una librería externa |
| `pip list` | Muestra los paquetes instalados |
| `help(objeto)` | Muestra la documentación de una función o módulo |
| `type(objeto)` | Muestra el tipo de dato de una variable |

En un notebook, estos mismos comandos se escriben en una celda y se ejecutan con Shift + Enter.

## Ventajas y desventajas

**A favor:** sintaxis simple, ecosistema muy amplio, comunidad grande y soporte nativo en Colab.

**En contra:** al ser interpretado, suele ser más lento que un lenguaje compilado en cómputo numérico intensivo (por eso NumPy usa código en C por debajo). El manejo de entornos (`pip`, `conda`, `venv`) puede confundir al principio.
