## Instalación

Python se puede usar de varias formas, según qué tan cómodo quieras tenerlo:

**Opción 1 — Instalación local desde python.org**
Se descarga el instalador desde [python.org](https://www.python.org/downloads/) según tu sistema operativo. Una vez instalado, puedes correrlo desde la terminal (`python`) o integrarlo a un editor como VS Code con la extensión oficial de Python.

**Opción 2 — Anaconda (recomendada si vas a usar ciencia de datos)**
[Anaconda](https://www.anaconda.com/download) instala Python junto con las librerías más usadas para cómputo científico (NumPy, pandas, Matplotlib) y el gestor de entornos `conda`, que evita conflictos entre proyectos que necesitan versiones distintas de una misma librería.

**Opción 3 — Jupyter (para trabajar con notebooks)**
Si ya tienes Python instalado (por cualquiera de las dos opciones anteriores), Jupyter se instala con:

```bash
pip install notebook
jupyter notebook
```

Esto abre Jupyter en tu navegador, donde puedes crear notebooks `.ipynb` con celdas de código y de texto.

**Opción 4 — Google Colab (sin instalar nada)**
[Colab](https://colab.research.google.com/) corre notebooks de Python directamente en el navegador, con los paquetes más comunes ya instalados y acceso gratuito a GPU. Es la forma más rápida de empezar si no quieres instalar nada en tu equipo — a diferencia de R y Julia, Python tiene soporte nativo completo en Colab.

### Ventajas y desventajas de Python

- A favor: sintaxis simple y legible, ideal para quien empieza a programar.
- A favor: ecosistema enorme de librerías (ciencia de datos, web, automatización, machine learning) y una comunidad muy grande, con mucha documentación.
- A favor: multiplataforma y con soporte nativo en Colab, sin instalar nada.
- En contra: al ser interpretado, suele ser más lento que lenguajes compilados para cómputo numérico intensivo (por eso existen NumPy y librerías en C/C++ por debajo, para compensar).
- En contra: el manejo de entornos y versiones (pip vs conda, virtualenv, etc.) puede ser confuso al principio.

## Comandos básicos

Estos son los comandos que vas a usar todo el tiempo, ya sea en la consola de Python, en una celda de Jupyter o en la terminal:

- `python archivo.py` — ejecuta un script de Python desde la terminal.
- `python` (sin argumentos) — abre el intérprete interactivo (REPL); cada línea se ejecuta al momento. Se sale con `exit()` o Ctrl+D.
- `pip install nombre_paquete` — instala una librería externa (por ejemplo `pip install numpy`).
- `pip list` — muestra los paquetes instalados en el entorno actual.
- `help(objeto)` — dentro del intérprete, muestra la documentación de una función, clase o módulo. Por ejemplo: `help(print)`.
- `type(objeto)` — muestra el tipo de dato de una variable; útil para depurar.

En un notebook (Jupyter, Colab), estos mismos comandos se escriben dentro de una celda de código y se ejecutan con Shift + Enter, igual que cualquier otro código de Python.
